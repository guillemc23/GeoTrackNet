```python
import tensorflow as tf
import pickle

with tf.gfile.Open('../../data/ct_2017010203_10_20/ct_2017010203_10_20_train.pkl', "rb") as f:
    Vs = pickle.load(f)

Vs
```

stats

```python
max_messages_amount = 0
min_messages_amount = 9999
for i in range(len(Vs)):
    mmsi_messages_amount = Vs[i].shape[0]
    if max_messages_amount < mmsi_messages_amount:
        max_messages_amount = mmsi_messages_amount
    if min_messages_amount > mmsi_messages_amount:
        min_messages_amount = mmsi_messages_amount

print(f'amount of MMSIs {len(Vs)}')
print(f'min amount of MMSI messages {min_messages_amount}')
print(f'max amount of MMSI messages {max_messages_amount}')
```
