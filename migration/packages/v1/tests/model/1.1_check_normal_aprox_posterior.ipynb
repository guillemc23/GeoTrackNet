```python
import tensorflow as tf
tf.compat.v1.enable_eager_execution()
```

```python
latent_size = 64
_DEFAULT_INITIALIZERS = {"w": tf.compat.v1.keras.initializers.VarianceScaling(scale=1.0, mode="fan_avg", distribution="uniform",seed=111),
                         "b": tf.compat.v1.zeros_initializer()}
```

```python
from migration.models.vrnn import NormalApproximatePosterior, ConditionalNormalDistribution

prior = ConditionalNormalDistribution(size=latent_size, hidden_layer_sizes=[latent_size], sigma_min=0, raw_sigma_bias=0.25, initializers=_DEFAULT_INITIALIZERS)
approx_posterior = NormalApproximatePosterior(size=latent_size, hidden_layer_sizes=[latent_size], sigma_min=0, raw_sigma_bias=0.25, initializers=_DEFAULT_INITIALIZERS)
```

```python
rnn_out = tf.random.uniform((32,64), dtype=tf.float32)
targets_encoded = tf.random.uniform((32,64), dtype=tf.float32)
```

```python
latent_dist_prior = prior(rnn_out)
```

```python
latent_dist_q = approx_posterior(rnn_out, targets_encoded, prior_mu= latent_dist_prior.loc)
```

```python
latent_state = latent_dist_q.sample(seed=1)  
latent_state_prior = latent_dist_prior.sample(seed=1)
```
