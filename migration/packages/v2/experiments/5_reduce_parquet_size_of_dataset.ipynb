```python
import pandas as pd
```

```python
for sub in ['test', 'train', 'val']:
    parquet = f'../new_data/ais_{sub}.parquet'
    _ds = pd.read_parquet(parquet).replace(1, 0.9999)
    _ds.insert(0, 'temp_id', range(0, len(_ds)))
    _ds = _ds.set_index('temp_id', append=True)
    _ds = _ds.sort_index().reset_index(1).drop('temp_id', axis=1)
    _ds.insert(0, 't', range(0, len(_ds)))
    _ds.groupby('t').first().rename(columns={'t': 't_0'})
    _ds = _ds.join(_ds.groupby('track_id').first().rename(columns={'t': 't0'})['t0'])
    _ds['t'] -= _ds['t0']
    _ds.drop('t0', axis = 1, inplace=True)
    _ds =_ds.astype({
        't' : 'uint16',
        'latitude' : 'float32',
        'longitude' : 'float32',
        'cog' : 'float32',
        'sog' : 'float32',
        }, copy=True)
    _ds = _ds.set_index('t',append=True)
    _ds.to_parquet(f'../new_data/ais_{sub}_arrow_multi_index.parquet', store_decimal_as_integer=True)
```
