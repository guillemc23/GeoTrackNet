```python
import pandas as pd
```

```python
for sub in  ['test', 'train', 'val']:
    parquet = f'../new_data/ais_{sub}_arrow.parquet'
    _ds = pd.read_parquet(parquet, columns=['latitude', 'longitude', 't'])
    _ds = _ds.set_index('t',append=True)
    _ds['latitude'] = (300 * _ds['latitude']).astype('uint16')
    _ds['longitude'] = (300 * _ds['longitude']).astype('uint16')
    _ds = _ds.rename({'latitude': 'lat_bin', 'longitude': 'lon_bin'})
    _ds.to_parquet(f'../new_data/ais_{sub}_by_bins.parquet')
```
