```python
import tensorflow as tf
import numpy as np
```

```python
latent_size = 64
_DEFAULT_INITIALIZERS = {"w": tf.keras.initializers.VarianceScaling(scale=1.0, mode="fan_avg", distribution="uniform",seed=111),
                         "b": tf.zeros_initializer()}
```

```python
from migration.models.distributions import ConditionalNormalDistribution


prior = ConditionalNormalDistribution(size=latent_size, hidden_layer_size=latent_size, sigma_min=0, raw_sigma_bias=0.25, initializers=_DEFAULT_INITIALIZERS)
```

```python
rnn_out = tf.random.uniform((32,64), dtype=tf.float32)
```

```python
latent_dist_prior = prior(rnn_out)
```
